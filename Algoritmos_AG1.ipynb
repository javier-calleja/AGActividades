##Torres de Hanoi con Divide y vencerás

```
def Torres_Hanoi(N, desde, hasta, cont):
  if N ==1 :
    print("Count ", cont, "Lleva la ficha FIN " ,desde , " hasta ", hasta )
  else:
    #Número mágico 6 para obtener el poste destino basado en el cálculo '1+2+3 = 6', por lo que sólo sirve para 3 postes: mejora,
    Torres_Hanoi(N-1, desde, 6-desde-hasta, cont+1)
    print("Count ", cont, "Lleva la ficha MID " ,desde , " hasta ", hasta)
    Torres_Hanoi(N-1, 6-desde-hasta  , hasta, cont+1 )


Torres_Hanoi(4, 1 , 3, 0)
```

```
#Sucesión_de_Fibonacci
#https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci
#Calculo del termino n-simo de la suscesión de Fibonacci
def Fibonacci(N:int):
  if N < 2:
    return 1
  else:
    return Fibonacci(N-1)+Fibonacci(N-2)

# Añadimos iteratividad para pasar a complejidad lineal, invertimos orden y almacenamos valores ya calculados
def Fibonacci_Iterativo(N:int):
  if N < 2:
    return 1
  else:
    a, b = 1, 1  # Valores iniciales F1 y F2
    for i in range(2, N+1):
      a, b = b, a+b
  return b


res = Fibonacci(5)
res_iter = Fibonacci_Iterativo(5)
print(f'Resultado Fibonacci: {res} Iterativo {res_iter}')
assert(res == res_iter)
```

##Devolución de cambio por técnica voraz

```
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION


cambio_monedas(15,[25,10,5,1])
```

## N-Reinas por técnica de vueta atrás


```
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True



def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0

reinas(4)
```

## Viaje por el rio. Programación dinámica

```
TARIFAS = [
[0,5,4,3,999,999,999],
[999,0,999,2,3,999,11],
[999,999, 0,1,999,4,10],
[999,999,999, 0,5,6,9],
[999,999, 999,999,0,999,4],
[999,999, 999,999,999,0,3],
[999,999,999,999,999,999,0]
]



################################################################
def Precios(TARIFAS):
################################################################
  #Total de Nodos
  N = len(TARIFAS[0])

  #Inicialización de la tabla de precios
  PRECIOS = [ [9999]*N for i in [9999]*N]
  RUTA = [ [""]*N for i in [""]*N]

  for i in range(0,N-1):
    RUTA[i][i] = i             #Para ir de i a i se "pasa por i"
    PRECIOS[i][i] = 0          #Para ir de i a i se se paga 0
    for j in range(i+1, N):
      MIN = TARIFAS[i][j]
      RUTA[i][j] = i

      for k in range(i, j):
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
            MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j] )
            RUTA[i][j] = k          #Anota que para ir de i a j hay que pasar por k
        PRECIOS[i][j] = MIN

  return PRECIOS,RUTA
################################################################


PRECIOS,RUTA = Precios(TARIFAS)
#print(PRECIOS[0][6])

print("PRECIOS")
for i in range(len(TARIFAS)):
  print(PRECIOS[i])

print("\nRUTA")
for i in range(len(TARIFAS)):
  print(RUTA[i])

#Determinar la ruta con Recursividad
def calcular_ruta(RUTA, desde, hasta):
  if desde == hasta:
    #print("Ir a :" + str(desde))
    return ""
  else:
    return str(calcular_ruta( RUTA, desde, RUTA[desde][hasta])) +  \
                ',' + \
                str(RUTA[desde][hasta] \
              )

print("\nLa ruta es:")
calcular_ruta(RUTA, 0,6)
```



**Reto propuesto: Encontrar los dos puntos más cercanos**

Dado un conjunto de puntos se trata de encontrar los dos puntos más cercanos Guía para aprendizaje:

* Suponer en 1D, o sea, una lista de números: `[3403, 4537, 9089, 9746, 7259, ....`
* Primer intento: Fuerza bruta
* Calcular la complejidad. ¿Se puede mejorar?
* Segundo intento. Aplicar Divide y Vencerás
* Calcular la complejidad. ¿Se puede mejorar?
* Extender el algoritmo a 2D: `[(1122, 6175), (135, 4076), (7296, 2741)...`
* Extender el algoritmo a 3D.

```

import math
#Fuerza bruta, se recorren todos los elementos de la lista para calcular distancias y quedarnos con el mínimo: Coste temporal O(n**2)
def EncontrarDistanciaMin (lista):
  min = 100000 #suponemos un valor máximo
  longitud = len(lista)
  for i in range(longitud):
    for j in range(i+1, longitud):
      distancia = abs(lista[i]-lista[j])
      if min > distancia:
        min = distancia
        menor = (lista[i], lista [j])
  return min, menor

lista = [3403, 4537, 9089, 9746, 7259]

val_min, puntos = EncontrarDistanciaMin(lista)
print(f'Distancia: {val_min} Puntos: {puntos}')

#Mejora: ordenar, se compara sólo con el siguiente en la lista, se pasa a coste temporal O(n)
def EncontrarDistanciaMinOrdenada (lista):
  min = 100000 #suponemos un valor máximo
  longitud = len(lista)
  for i in range(longitud-1):
    distancia = abs(lista[i]-lista[i+1])
    if min > distancia:
      min = distancia
      menor = (lista[i], lista [j])
  return min, menor


val_min, puntos = EncontrarDistanciaMin(sorted(lista))
print(f'Distancia: {val_min} Puntos: {puntos}')


#Aplicando divide y venceras a lista ordenada se pasa a O(n*log(n))
def EncontrarDistanciaMinDivideVenceras(lista):
    long_lista = len(lista)

    # Caso base: para 3 o menos elementos
    if long_lista <= 3:
        min_dist = 100000
        menor = (None, None)
        for i in range(long_lista):
            for j in range(i + 1, long_lista):
                distancia = abs(lista[i]-lista[j])
                if distancia < min_dist:
                    min_dist = distancia
                    menor = (lista[i], lista[j])
        return min_dist, menor

    # División
    medio = long_lista // 2
    izq_lista = lista[:medio]
    der_lista = lista[medio:]

    # Recursividad
    min_izq, menor_izq = EncontrarDistanciaMinDivideVenceras(izq_lista)
    min_der, menor_der = EncontrarDistanciaMinDivideVenceras(der_lista)

    # Comparación entre las 2 mitades
    if min_izq < min_der:
        min_dist = min_izq
        menor = menor_izq
    else:
        min_dist = min_der
        menor = menor_der

    # Combinación: se evalúa el borde derecho de la izq y el borde izquierdo de la der
    if len(izq_lista) > 0 and len(der_lista) > 0:
        frontera_izq = izq_lista[-1]  # Último elemento de la izquierda
        frontera_der = der_lista[0]   # Primer elemento de la derecha
        dist_frontera = abs(frontera_der - frontera_izq)

        if dist_frontera < min_dist:
            min_dist = dist_frontera
            menor = (frontera_izq, frontera_der)

    return min_dist, menor

val_min, puntos = EncontrarDistanciaMinDivideVenceras(sorted(lista))
print(f'Distancia: {val_min} Puntos: {puntos}')


# Para 2 dimensiones y 3 dimensiones, se deberá aplicar la distancia euclidea (math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)) sobre el mismo algoritmo
def distancia_euclidea(p1, p2):
  return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)

#Utilizamos fuerza bruta porque para ordenar la lista por distancia euclidea ya se pierde la eficiencia ganada en divide y vencerás
def EncontrarDistanciaMin2D (lista):
  min = 100000 #suponemos un valor máximo
  longitud = len(lista)
  for i in range(longitud):
    for j in range(i+1, longitud):
      distancia = distancia_euclidea(lista[i], lista[j])
      if min > distancia:
        min = distancia
        menor = (lista[i], lista [j])
  return min, menor

lista2D = [(1122, 6175), (135, 4076), (7296, 2741)]
val_min, puntos2D = EncontrarDistanciaMin2D(lista2D)
print(f'Distancia: {round(val_min,2)} Puntos: {puntos2D}')
```
